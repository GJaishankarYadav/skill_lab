```python
import os
from getpass import getpass

os.environ["GEMINI_API_KEY"] = getpass("Enter Gemini API Key: ")
```

```python
import asyncio
from google import genai
from google.genai import types
from playwright.async_api import async_playwright

client = genai.Client()

print("All libraries imported successfully!")
```

```python
async def start_browser():

    global pw, browser, context, page

    pw = await async_playwright().start()

    browser = await pw.chromium.launch(
        headless=True
    )

    context = await browser.new_context()

    page = await context.new_page()

    print("Chrome browser is ready!")

await start_browser()

```

```python
# Clean up previous failed Playwright session
try:
    await pw.stop()
except:
    pass

# Start browser again
await start_browser()
```

```python
async def open_amazon():

    print("Opening Amazon India...")

    await page.goto(
        "https://www.amazon.in",
        wait_until="domcontentloaded",
        timeout=60000
    )

    print("Amazon opened successfully!")

    print("Website title:", await page.title())
    print("Website URL:", page.url)


await open_amazon()
```

```python
async def search_product(keyword: str):

    print(f"Searching for: {keyword}")

    try:
        search_box = page.locator("#twotabsearchtextbox")

        await search_box.wait_for(
            state="visible",
            timeout=15000
        )

        await search_box.fill(keyword)

        await page.locator(
            "#nav-search-submit-button"
        ).click()

        await page.wait_for_selector(
            "[data-component-type='s-search-result']",
            timeout=20000
        )

        print("Product search completed!")

        return True

    except Exception as e:
        print("Search failed:", str(e))
        return False


result = await search_product("bluetooth headphones")

print("Search status:", result)
```
